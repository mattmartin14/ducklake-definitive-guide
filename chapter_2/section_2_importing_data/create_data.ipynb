### Creating A CSV File for Import Demo
This script simply creates a csv file that we will import to ducklake

```python
import csv

rows = [
    {"order_id": 1001, "order_date": "2026-03-01", "customer_id": "C001"},
    {"order_id": 1002, "order_date": "2026-03-02", "customer_id": "C002"},
    {"order_id": 1003, "order_date": "2026-03-03", "customer_id": "C003"},
]

with open("customer_orders.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["order_id", "order_date", "customer_id"])
    writer.writeheader()
    writer.writerows(rows)

print('Created "customer_orders.csv" with 3 rows.')
```

```python
with open("customer_orders.csv", "w") as f:
    f.write("""order_id,order_date,customer_id
1001,2026-03-01,C001
1002,2026-03-02,C002
1003,2026-03-03,C003
""")
```
